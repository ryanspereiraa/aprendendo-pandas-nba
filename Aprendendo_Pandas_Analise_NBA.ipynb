# 1- Introdução ao Pandas e Estruturas de Dados
## O que é o Pandas
O Pandas é a principal biblioteca do Python para manipulação e análise de dados. Ele transforma dados brutos em tabelas estruturadas e permite fazer cálculos, estatísticas e limpezas com poucas linhas de código.
### Estruturas fundamentais:
1. **Series:** É uma coluna individual de dados (uma lista vertical com um índice).
2. **DataFrame:** É a tabela completa de 2 dimensões (linhas e colunas), onde cada coluna é uma *Series*.

### Detalhes dos comandos fundamentais

* **import pandas as pd:** Importa a biblioteca e define o apelido pd. Essa abreviação é a convenção universal adotada no mercado de tecnologia.
* **pd.DataFrame():** É o construtor da biblioteca. Ele pega uma estrutura de dados do Python e a converte formalmente em uma tabela de duas dimensões, atribuindo índices automáticos.
* **display() e print():**
 * print()
 : Exibe os dados como não formatado, útil para saídas simples no terminal.
 * display()
 : Função  que renderiza a tabela em HTML formatado, facilitando a leitura visual com linhas alternadas e destaque nos cabeçalhos.

```python
import pandas as pd
# criando uma series
pontos_edwards = pd.Series([28, 31, 19, 25], name = 'pontos')
print('Exemplo de uma Series:')
print(pontos_edwards)
print('\n' + '='*40 + '\n')

# criando um dataframe
dados_df = {
    'jogador': ['Edwards', 'Edwards', 'James', 'James'],
    'time': ['MIN', 'MIN', 'LAL', 'LAL'],
    'pontos': [28 , 31, 22, 29],
    'assistencias': [5, 7, 4, 6]
}

df_dados = pd.DataFrame(dados_df)
print('Exemplo de um DataFrame:')
display(df_dados)
```

## Inspeção Inicial de Dados
Quando se recebe uma base de dados nova, é preciso entender o tamanho dela, o tipo de informação em cada coluna e se há erros.

* **.head():** Mostra as primeiras linhas da tabela.
* **.tail():** Mostra as ultimas linhas da tabela
* **.info():** Exibe os tipos de dados (números, textos) e se há dados faltantes.
* **.describe():** Gera um resumo estatístico rápido (média, mínimo, máximo e etc.).

```python
# Visualizando as primeiras linhas
print('Visualizando com .head():')
display(df_dados.head(2))

# Visualizando as ultimas linhas
print('Visualizando com .tail():')
display(df_dados.tail(2))

# Verificando a estrutura técnica
print('\n Informações da tabela com .info():')
df_dados.info()

# Gerando estatísticas descritivas básicas
print('\n Estatísticas com .describe():')
display(df_dados.describe())
```
